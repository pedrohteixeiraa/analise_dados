```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
t2012 = pd.read_csv(
    "Tabela5-sem_emprego_2012.csv",
    sep=";",
    decimal=","
)

t2012
```

```python
t2026 = pd.read_csv(
    "Tabela5-sem_emprego_2026.csv",
    sep=";",
    decimal=","
)

t2026
```

```python
t2012 = t2012.rename(
    columns={
        "Desocupados - mulheres (2012 T1)": "mulheres_2012"
    }
)
```

```python
t2026 = t2026.rename(
    columns={
        "Desocupados - mulheres (2026 T1)": "mulheres_2026"
    }
)
```

```python
comp = t2012.merge(
    t2026,
    on=["Sigla", "Código", "Estado"],
    how="inner"
)

comp
```

```python
path = r"H:\Phyton\3° ETAPA\indicador_mulher\Tabela 1.1.1.xls"
indicador_1 = pd.read_excel(path, engine="xlrd")
indicador_1
```

```python
# Lê sem usar a 1ª linha como nome de coluna (senão o título vira "coluna")
bruto = pd.read_excel(path, engine="xlrd", header=None)

# Linhas 0–7: título/cabeçalho  |  8–40: resultados  |  41+: fonte e notas
indicador_1 = bruto.iloc[8:41].copy()
indicador_1.columns = [
    "uf_regiao",
    "total",
    "total_branca",
    "total_preta_parda",
    "homem_branca",
    "homem_preta_parda",
    "mulher_branca",
    "mulher_preta_parda",
]
indicador_1 = indicador_1.reset_index(drop=True)

# horas são número, não texto
for col in indicador_1.columns[1:]:
    indicador_1[col] = pd.to_numeric(indicador_1[col], errors="coerce")

indicador_1
```

```python
comp = comp.merge(
    indicador_1[["uf_regiao", "mulher_branca"]],
    left_on="Estado",
    right_on="uf_regiao",
    how="inner"
)

comp
```

```python
ordem = comp.sort_values(
    "mulheres_2026",
    ascending=False
)["Estado"]
```

```python
longo = comp.melt(
    id_vars=["Estado", "mulher_branca"],
    value_vars=["mulheres_2012", "mulheres_2026"],
    var_name="Ano",
    value_name="Participacao_mulheres"
)
```

```python
longo["Ano"] = longo["Ano"].map({
    "mulheres_2012": "2012 T1",
    "mulheres_2026": "2026 T1"
})

longo
```

```python
fig, ax = plt.subplots(figsize=(10, 10))

sns.barplot(
    data=longo,
    y="Estado",
    x="Participacao_mulheres",
    hue="Ano",
    order=ordem,
    ax=ax,
)

ax.axvline(
    50,
    color="gray",
    linestyle="--",
    linewidth=1
)

ax.set_xlabel(
    "Participação das mulheres entre as pessoas desocupadas (%)"
)

ax.set_ylabel("")

ax.set_title(
    "Desocupação: participação feminina em 2012 T1 e 2026 T1"
)

ax.legend(title="Ano")

fig.tight_layout()
plt.show()
```
