```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
brasil = pd.DataFrame(
    {
        "area": [
            "Graduação (total)",
            "STEM",
            "Educação / Saúde"
        ],
        "homens": [
            dados_gerais.loc[
                dados_gerais["uf_regiao"] == "Brasil",
                "homens"
            ].iloc[0],

            dados_matematica.loc[
                dados_matematica["uf_regiao"] == "Brasil",
                "homens"
            ].iloc[0],

            dados_educacao.loc[
                dados_educacao["uf_regiao"] == "Brasil",
                "homens"
            ].iloc[0]
        ],
        "mulheres": [
            dados_gerais.loc[
                dados_gerais["uf_regiao"] == "Brasil",
                "mulheres"
            ].iloc[0],

            dados_matematica.loc[
                dados_matematica["uf_regiao"] == "Brasil",
                "mulheres"
            ].iloc[0],

            dados_educacao.loc[
                dados_educacao["uf_regiao"] == "Brasil",
                "mulheres"
            ].iloc[0]
        ]
    }
)
```

```python
brasil_long = brasil.melt(
    id_vars="area",
    var_name="sexo",
    value_name="pessoas"
)

brasil_long
```

```python
fig, ax = plt.subplots(figsize=(9, 5))

sns.barplot(
    data=brasil_long,
    x="area",
    y="pessoas",
    hue="sexo",
    ax=ax
)

ax.set_title("Brasil: pessoas com graduação, por área e sexo")
ax.set_xlabel("Área")
ax.set_ylabel("Pessoas")

plt.tight_layout()
plt.show()
```
